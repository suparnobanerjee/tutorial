# Handling exceptions

```python
try:
  x=int(input("Enter 1st num: "))
  y=int(input("Enter 2nd num: "))
  print("x/y = ", x/y)
except ValueError:
  print("Please enter valid numbers!")
except ZeroDivisionError:
  print("You cannot divide a number by zero!")
finally:
  print("Done!")
```

# File Streams
### Lifecycle:
1. Opening the stream to req OS file descriptor.
2. Transferring bytes or characters via the stream cursor.
3. Closing the stream to flush buffered data to disk and free OS resources.

### Reading from a stream:
1. Line by line iteration (memory efficient esp for Gb-scale files)

```python
with open("data.txt", "w", encoding="utf-8") as stream:
    stream.write("First line of text\n")
    stream.write("Second line of text\n")

with open("data.txt", "r", encoding="utf-8") as stream:
    for line in stream:
        print(line.rstrip())
```

2. Chunk based binary or Fixed size reads

```python
with open("data.txt", "rb") as stream:
  while chunk := stream.read(1024):
    # Process 1024 characters at a time
    print("Raw bytes:", chunk)
    print("Decoded string:", chunk.decode("utf-8").rstrip())
```

3. Reading everything into a single memory variable

```python
with open("data.txt","r", encoding="utf-8") as stream:
  print(stream.read())
```

### Writing and Appending:
Writing to a stream places data in an internal memory buffer before the OS commits it to physical storage.

```python
# Overwrite mode
with open("log.txt","w", encoding="utf-8") as stream:
  stream.write("Execution started\n")
  stream.writelines(["Entry 1\n","Entry 2\n"])
```

```python
# Append mode
with open("log.txt", "a", encoding="utf-8") as stream:
    stream.write("New entry added later\n")

with open("log.txt","r", encoding="utf-8") as stream:
  print(stream.read())
```

# OS methods

1. os.getcwd(): Where am I?

```python
import os
print(os.getcwd())
```

2. os.listdir(): What files are here?

```python
print(os.listdir("."))
```

3. os.path.join(): Builds file paths safely.

```python
folder = "dataset"
filename = "dataset.txt"

print(os.path.join(folder, filename))
```

4. Check if a file exists

```python
if os.path.exists("data.txt"):
  print("File exists!")
else:
  print("File doesnot exists!")
```

5. Create Folders

```python
os.makedirs("output/logs", exist_ok=True)
# Using exist_ok=True prevents errors if the folder already exists.

print(os.listdir("."))
print(os.listdir("output"))
```

6. Delete a file or empty directory

```python
if os.path.exists("data.txt"):
  os.remove("data.txt")

if os.path.exists("input"):
  os.rmdir("input")

print(os.listdir("."))
```

7. Rename or Move files

```python
# Rename
os.rename("data_backup.txt", "data.txt")

# Move to another directory
os.rename("data.txt", "output/data.txt")

print(os.listdir("."))
print(os.listdir("output"))
```

# String functions

1. Clean edges, Change casing

```python
input = "  Hello\n"
print(input.strip())
print(input.strip().lower())
print(input.strip().upper())
print(input.strip().title())

# Common check to avoid case sensitivity
if input.strip().lower() == "hello":
    print("Match found")
```

2. Break apart, Glue together

```python
data = "a,b,c"
print(data.split(","))

print("-".join(data.split(",")))  # Glue with hyphen
print(", ".join(data.split(","))) # Glue with comma and space
```
